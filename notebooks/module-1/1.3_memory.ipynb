## No memory

```python
from langchain.agents import create_agent


agent = create_agent(
    "gpt-5-nano"
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Mr.Bot and my favourite colour is Green.")

response = agent.invoke(
    {"messages": [question]} 
)
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")
# Without the memory saver, the agent will not remember the previous conversation and will not be able to answer the question correctly.
response = agent.invoke(
    {"messages": [question]} 
)

pprint(response)
```

## Memory

```python
from langgraph.checkpoint.memory import InMemorySaver  
#using memory saver to store the state of the agent in memory, so that it can be restored later

agent = create_agent(
    "gpt-5-nano",
    checkpointer=InMemorySaver(),  
)
```

```python
from langchain.messages import HumanMessage
# Using a different name and favourite colour to demonstrate that the agent can remember the previous conversation and answer the question correctly.
question = HumanMessage(content="Hello my name is LangChainBot and my favourite colour is Green.")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,  
)
```

```python
pprint(response)
# Now, let's ask the agent about its favourite colour again to see if it remembers the previous conversation.
```

```python
question = HumanMessage(content="What's my favourite colour?")
# With the memory saver, the agent should remember the previous conversation and answer the question correctly.
response = agent.invoke(
    {"messages": [question]},
    config,  
)

pprint(response)
```
