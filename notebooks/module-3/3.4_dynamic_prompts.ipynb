```python
from dotenv import load_dotenv

load_dotenv()
```

```python
from dataclasses import dataclass
from langchain.agents.middleware import dynamic_prompt, ModelRequest

# Define a dataclass to hold the user's language preference. This will be used to dynamically generate prompts in the user's preferred language.
# LanguageContext is a simple dataclass that holds the user's language preference. It defaults to "English" but can be set to any other language as needed.

@dataclass
class LanguageContext:
    user_language: str = "English"

# Dynamic prompt function that generates a system prompt based on the user's language preference. 
# If the user prefers a language other than English, the prompt will instruct the assistant to respond only in that language. Otherwise, it will use the default English prompt.
# Dynamic prompts are useful for creating a more personalized experience for users, allowing the assistant to communicate in the user's preferred language.
@dynamic_prompt
def user_language_prompt(request: ModelRequest) -> str:
    """Generate system prompt based on user role."""
    user_language = request.runtime.context.user_language
    base_prompt = "You are a helpful assistant."

    if user_language != "English":
        return f"{base_prompt} only respond in {user_language}."
    elif user_language == "English":
        return base_prompt
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model="gpt-5-nano", #gpt-5-nano is a smaller model with a smaller context window, but it is more efficient for shorter conversations.
    context_schema=LanguageContext,
    middleware=[user_language_prompt]
)
```

```python
from langchain.messages import HumanMessage

response = agent.invoke(
    {"messages": [HumanMessage(content="Hello, how are you?")]},
    context=LanguageContext(user_language="Irish") # Set the user's language preference to Irish, which will trigger the dynamic prompt to instruct the assistant to respond only in Irish.
)

print(response["messages"][-1].content)
```
